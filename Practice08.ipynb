create a student clasee that takes name and marks of 3 subjects as arguments in contructor .then create a method to print the average

```python
class student:
    def __init__(self, name, marks):  # Fixed: Changed _init_ to __init__ (double underscores)
        self.name = name
        self.marks = marks

    def get_avg(self):
        sum = 0
        for i in self.marks:
            sum += i
        print('congrats your Average in 3 subjects is ', sum/3)


s1 = student('Fawad ahmad', [99, 98, 100])
s1.get_avg()
```
